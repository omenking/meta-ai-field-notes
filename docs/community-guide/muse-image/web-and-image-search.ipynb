```python
import base64
import os
import requests
from pathlib import Path
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import display, Image, HTML
```

```python
load_dotenv("../.env") # load our env vars
client = OpenAI(
    base_url="https://api.meta.ai/v1",
    api_key=os.environ["MODEL_API_KEY"],
)
```

```python
response = client.responses.create(
    model="muse-image-1.0",
    input=(
        "Search the web and image search for Canadian Pacific "
        "locomotive CP 8171, a GE AC4400CWM photographed at "
        "Schreiber, Ontario. Study photographs of this exact locomotive, "
        "including its cab, nose, trucks, roof equipment, paint scheme, "
        "lettering and road number. Research its associated grain train. "
        "Create a historically and mechanically accurate freight train "
        "scene passing through Schreiber. Render it in a foggy PS1-era "
        "video game style, preserving the actual locomotive proportions "
        "and recognizable railway details. Do not invent locomotive "
        "features or substitute a different model."
    ),
    tools=[{
        "type": "image_generation",
        "size": "1536x1024",
        "enable_web_search": True,
        "enable_image_search": True
    }]
)
display(Image(data=base64.b64decode(next(x.result for x in response.output if x.type == "image_generation_call")), width=500))
```
